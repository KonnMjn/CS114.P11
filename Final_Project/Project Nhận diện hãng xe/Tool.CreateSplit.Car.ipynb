```python
import os
import csv
import random
from collections import defaultdict
```

```python
# Định nghĩa CategoryID cho các hãng xe
CATEGORY_MAPPING = {
    "Honda": 1,
    "Hyundai": 2,
    "KIA": 3,
    "Mazda": 4,
    "Mitsubishi": 5,
    "Suzuki": 6,
    "Toyota": 7,
    "VinFast": 8,
    "Others": 0
}
```

```python
def scan_images(directory):
    """
    Quét thư mục cha để tìm tất cả ảnh .jpg và gán CategoryID.
    """
    dataset = set()  # Sử dụng set để loại bỏ trùng lặp
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith('.jpg'):
                # Xác định hãng xe từ tên thư mục
                category = os.path.basename(root)
                category_id = CATEGORY_MAPPING.get(category, CATEGORY_MAPPING["Others"])
                # Đường dẫn ảnh đầy đủ
                relative_path = os.path.join(category, file)
                dataset.add((relative_path, category_id))  # Thêm vào set
    return list(dataset)  # Chuyển set về list để xử lý tiếp


def save_to_csv(data, filename):
    """ 
    Ghi dữ liệu vào file CSV với header "ImageFullPath,CategoryID".
    """
    with open(filename, mode='w', newline='', encoding='utf-8') as file:
        writer = csv.writer(file)
        # Ghi header vào file
        writer.writerow(["ImageFullPath", "CategoryID"])
        for row in data:
            writer.writerow(row)

def create_splits(data, num_splits=5):
    """
    Tạo các tập Split Train/Test từ dữ liệu CarDataset.csv.
    """
    category_groups = defaultdict(list)
    for image_path, category_id in data:
        category_groups[category_id].append(image_path)

    splits = {i: [] for i in range(1, num_splits + 1)}
    for category_id, images in category_groups.items():
        random.shuffle(images)
        fold_size = len(images) // num_splits
        folds = [images[i * fold_size:(i + 1) * fold_size] for i in range(num_splits)]
        for i, fold in enumerate(folds, start=1):
            splits[i].extend([(img, category_id) for img in fold])

    return splits

def save_splits(splits, output_prefix="CarDataset-Splits"):
    """
    Ghi các tập Train/Test vào file CSV.
    """
    for i in range(1, 6):
        train_data = []
        test_data = splits[i]
        for j in range(1, 6):
            if j != i:
                train_data.extend(splits[j])

        # Ghi Train
        train_filename = f"{output_prefix}-{i}-Train.csv"
        save_to_csv(train_data, train_filename)

        # Ghi Test
        test_filename = f"{output_prefix}-{i}-Test.csv"
        save_to_csv(test_data, test_filename)

        # Thống kê và in kết quả
        print(f"\nSplit-{i} Train:")
        print(f"  File: {train_filename}")
        print(f"  Total Images: {len(train_data)}")
        category_counts = defaultdict(int)
        for _, category_id in train_data:
            category_counts[category_id] += 1
        for category_id, count in category_counts.items():
            print(f"    Category {category_id}: {count} images")

        print(f"\nSplit-{i} Test:")
        print(f"  File: {test_filename}")
        print(f"  Total Images: {len(test_data)}")
        category_counts = defaultdict(int)
        for _, category_id in test_data:
            category_counts[category_id] += 1
        for category_id, count in category_counts.items():
            print(f"    Category {category_id}: {count} images")
```

```python
# Đường dẫn thư mục cha
root_directory = "D:\\Dataset\\FinalData_CS114"

# Bước 1: Quét thư mục để tạo file CarDataset.csv
dataset = scan_images(root_directory)
save_to_csv(dataset, "CarDataset.csv")
print("\nFile CarDataset.csv đã được tạo!")

# Bước 2: Tạo các splits và ghi vào file
splits = create_splits(dataset)
save_splits(splits)
```
